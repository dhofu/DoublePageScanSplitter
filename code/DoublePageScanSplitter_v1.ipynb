```python
import os
import cv2
import numpy as np
import time
import pandas as pd
from datetime import datetime
```

```python
# get the starting time of the process
st = time.time()
```

```python
sourceDir = r"..\data\images" # update source directory
targetDir = r"..\data\split" # update target directory
```

```python
filepaths = []
for subdir, dirs, files in os.walk(sourceDir):
    for file in files:
        if file.lower().endswith('.jpg'):
            sourceFilepath = os.path.join(subdir, file)
            filepaths.append(sourceFilepath)
        else:
            print('not a jpg')
```

```python
i = len(filepaths)
print(i)
```

```python
def click_event(event, x, y, flags, param):
    if event == cv2.EVENT_LBUTTONDOWN:
        print(x)
        xs.append(x)
        font = cv2.FONT_HERSHEY_SIMPLEX
        strX = str(x)
        cv2.putText(img, strX, (x,y), font, 1.5, (255,0,255), 5)
        cv2.imshow("Resized_Window", img)
    if event == cv2.EVENT_RBUTTONDOWN:
        print('None')
        xs.append(0)
    if event == cv2.EVENT_LBUTTONUP:
#        cv2.waitKey(0)
        cv2.destroyAllWindows()
    if event == cv2.EVENT_RBUTTONUP:
#        cv2.waitKey(0)
        cv2.destroyAllWindows()    
```

```python
xs = [] # split value for x-axis

for filepath in filepaths:
    cv2.namedWindow("Resized_Window", cv2.WINDOW_NORMAL)
    cv2.resizeWindow("Resized_Window", 800, 600)
    img = cv2.imread(filepath)
    cv2.imshow("Resized_Window", img)
    cv2.setMouseCallback("Resized_Window", click_event)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
i = 0
while i < len(filepaths)-1:
    for subdir, dirs, files in os.walk(sourceDir):
        for file in files:
            if file.lower().endswith('.jpg'):
                sourceFilepath = os.path.join(subdir, file)                
                destFilepath = os.path.join(subdir, file).replace(sourceDir, targetDir)
                os.makedirs(os.path.dirname(destFilepath), exist_ok=True)
                print(sourceFilepath) # shows progress; can be commented out
                if xs[i] > 0:
                    img = cv2.imread(sourceFilepath)
                    img.shape
                    h = img.shape[0]
                    w = img.shape[1]
                    split = xs[i]
                    img_verso = img[0:h, 0:split]
                    img_recto = img[0:h, split:w]
                    cv2.imwrite(destFilepath.replace('.jpg', '_L_verso.jpg'), img_verso) # can be commented out if necessary
                    cv2.imwrite(destFilepath.replace('.jpg', '_R_recto.jpg'), img_recto) # can be commented out if necessary
                    i += 1
                else:
                    i += 1
```

img.shape returns (Height, Width, Number of Channels) <br>
where <br>
Height represents the number of pixel rows in the image or the number of pixels in each column of the image array. <br>
Width represents the number of pixel columns in the image or the number of pixels in each row of the image array.<br>
Number of Channels represents the number of components used to represent each pixel.<br>

Syntax for slicing a Numpy array.
cropped = img[start_row:end_row, start_col:end_col] <br>
with our variables, this means: img[0:h, 0:xs] for left side of the double page, and <br>
img[0:h, xs:w] for the right side of the double page.

```python
# get the end time
et = time.time()

# get the execution time
elapsed_time = et - st
strftime = time.strftime("%H:%M:%S", time.gmtime(elapsed_time))
print('Execution time:', elapsed_time, 'seconds')
print('Execution time:', time.strftime("%H:%M:%S", time.gmtime(elapsed_time)))

# last thing to add would be to append execution to a document with information about folder and time.
```

```python
# count number of files in targetDir using list comprehension
targetDirFileCount = len([entry for entry in os.listdir(targetDir) if os.path.isfile(os.path.join(targetDir, entry))])
print(targetDirFileCount)
```

```python
# datetime object containing current date and time
now = datetime.now()
print("now =", now)

# dd/mm/YY H:M:S
dt_string = now.strftime("%d/%m/%Y %H:%M:%S")
print("date and time =", dt_string)
```
